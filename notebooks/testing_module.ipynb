```python
#importing data
from doughflow.data.make_dataset import load_raw_data, get_data_summary, rename_data, aggregate_daily_level
```

```python
# loading data
df = load_raw_data("../data/raw")

new_df = rename_data(df)

new_df = aggregate_daily_level(new_df)

new_df
```

```python
from doughflow.features.build_features import TemporalFeatureExtractor, ProductFeatureExtractor, LagFeatureExtractor
from sklearn.pipeline import Pipeline

pipeline = Pipeline([
    ('product', ProductFeatureExtractor(aggregation_level='item')),
    ('temporal', TemporalFeatureExtractor()),
    ('lag', LagFeatureExtractor(
        recent_lag_weeks=[1,2,3,4],
        yearly_lags=[1,2],
        fill_strategy= 'cascading'
    ))
])

df_cascading = pipeline.fit_transform(new_df)

df_cascading
```
